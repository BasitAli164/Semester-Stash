Question: 
    Create a one-dimensional array in a program, print its type and dimension as well and then also show a list in your program.

```python
import numpy as np; # import numpy and create alias for easy to use
```

Array Part

```python
arr=np.array([1,2,3,4,5,6,7])
print(f"Array: {arr}")
print(f"Type of Array: {type(arr)}")
print(f"Dimension of Array: {arr.ndim}")
```

List Part

```python
list=[1,2,3,4,5,6,7]
print(f"List:{list}")
print(f"Type of list: {type(list)}")
# print(f"Dimension of list: {list.ndim}") # it is a list , so does n't find the dimension
```
