Question:
        Create a 2 D matrix of ones then also show its shape, size and data type and then change the datatype from float to integer.

```python
import numpy as np; # import numpy as set alias for easy to use
```

Method-1: To create 2D matrix 

```python
twoDMatrix=np.array([[1.,1,1,1],[1,1,1,1]]) # if there only one element has float then the  matrix datatype  will be float datatype not int 
print(f"2D Matrix is: {twoDMatrix}")
print(f"Shap of 2d Matrix:{twoDMatrix.shape}") # shap method return the number of row and column in matrix like this form (rows,column)
print(f"Size of 2d Matrix:{twoDMatrix.size}") # size method return the total number of element inside the matrix
print(f"Data Type of 2d Matrix:{twoDMatrix.dtype}") # dtype method return the data type of the matrix 
twoDMatrix.dtype=int # here we done typeCasting to change the data type from flaot to int
print(f"Change the Data Type of 2d Matrix:{twoDMatrix.dtype}")
```

Method-2: Create 2D matrix

```python
# towDMatix1=np.ones((2,3)) # Here we use two methods ones and shap. shap (2,3) this return 2 by 3 matrix and ones method insert 1 as element of the 2 by 3 matrix and by default it return float as datatype
towDMatix1=np.ones((2,3),dtype=int) # second argument change the dafualt datatype float to int
print(f"2D Matrix is: {towDMatix1}")
print(f"Shap of 2d Matrix:{towDMatix1.shape}") # shap method return the number of row and column in matrix like this form (rows,column)
print(f"Size of 2d Matrix:{towDMatix1.size}") # size method return the total number of element inside the matrix
print(f"Data Type of 2d Matrix:{towDMatix1.dtype}") # dtype method return the data type of the matrix 
```
