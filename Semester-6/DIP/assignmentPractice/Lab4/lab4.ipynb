### Write a program to apply some operations on string values.

```python
import numpy as np
```

```python
str1="Computer"
str2="Science"

joinBoth=np.char.add(str1,str2) # join both
print(f'Convert into Lower Case: {np.char.lower(joinBoth)}') # convert into lower case
print(f'Convert into Upper Case: {np.char.upper(joinBoth)}') # convert into upper case
print(f'Title :{np.char.title(joinBoth)}') # 
print(f'Center: {np.char.center(joinBoth,33)}') # Center
print(f'Center: {np.char.center(joinBoth,33,fillchar="-")}') # Center
# print(f'Splite: {np.char.split(joinBoth)}') # Does not work 
# print(f'SpliteLine:{np.char.splitlines("Computer\nScience")}') # Does not works

type(joinBoth)
```

```python
str3="DIP"
print(f'Join Method: {np.char.join(["."],[str3])}')
print(f'Replace Method:{np.char.replace}')
```
