# Assignment-3

Question:  
         Write a program then use some numpy functions or method in it.

```python
# arange function
np_arr1=np.arange(1,100,10) # arange return a vector or 1D array with stopValue-1 number of element
# np.arange(startValue,stopValue,steps or leave element how we want )
print("Array with Arange method: ",np_arr1)

```

```python
# linspace function or method
np_arr2=np.linspace(100,150,2) # linspace(start, stop, how many element want in array)
print("Array with linspace method:",np_arr2)
```

```python
# reshape method 
np_arr3=np.arange(1,21) # create a vector
print("Vector is: ",np_arr3)

reshapeArray=np.reshape(np_arr3,(2,10)) # convert the shape of vector into matrix
print(f"Reshape Array is:{reshapeArray}")



flattenArray=reshapeArray.flatten() # convert matrix into vector  , work with copy of the original array 
revalArray=reshapeArray.ravel() # convert matrix into vector , work with original array

print(f"Flatten Array:{flattenArray}\nReval Array:{revalArray}")



transpose=reshapeArray.T # take transpose using T property
print(f"Transpose: {transpose}")
```

# Assignment-4

Question:
        Perform some mathematical operations on matrices

```python
# create two matrix 
matrixA=np.arange(1,13).reshape(3,4)
matrixB=np.arange(13,25).reshape(3,4)

print(f"Matrix A:\n{matrixA}\nMatrix B:\n{matrixB}") # print both matrix
```

### Addition of two matrix

```python
additionOfMatrix=np.add(matrixA,matrixB)
print(f"Addition of tow matrix is:\n{additionOfMatrix}")
```

### Subtraction of tow matrix

```python
subtractionOfMatrix=np.subtract(matrixB,matrixA)
print(f"Subtraction of tow matrix is:\n{subtractionOfMatrix}")
```

### Multiplication of tow matrix

```python
# Method-1
multiplicationOfMatrix=np.multiply(matrixA,matrixB)
print(f"Multiplication of tow matrix is:\n{multiplicationOfMatrix}")
# Method-2
# print("matrixA * matrixB",matrixA@matrixB) # show error
```

### Division of Matrix

```python
divisionOfMatrix=np.divmod(matrixB,matrixA)
print(f"Multiplication of tow matrix is:\n{divisionOfMatrix}")
```

### Min and Max Value in Matrix

```python
print(matrixA)
print(f"Min Element in Matrix A:{matrixA.min()}\nMax Element in Matrix A:{matrixA.max()}") 
print(f"Max  Column  in Matrix A:{matrixA.max(axis=0)}\n Min  Row in Matrix A:{matrixA.max(axis=1)}")
```

### Sum operation 

```python
sumOfMatrix=np.sum(matrixA)
print(f"Sum of all element of matrix A:{sumOfMatrix}")


sumOfSingleRow=np.sum(matrixA,axis=1)
print(f"Sum of Single Row of Matrix A:{sumOfSingleRow}")



sumOfSingleCol=np.sum(matrixA,axis=0)
print(f"Sum of Single Col of Matrix A:{sumOfSingleCol}")
```

## Spliting Operation on Array

```python
# split on 1D array
np_arr7=np.arange(1,19)
print(f"Size of Array: {np_arr7}")


# split vector into
splitArr=np.split(np_arr7,6) 
print(f"Splite Array into 6:\n",splitArr)


# reshaping the vector to apply the vertical and horizontal split
reshapeArr=np_arr7.reshape(3,6)
print("Reshape the array:\n",reshapeArr)

# split the matrix
splitArr1=np.split(reshapeArr,3)
print("\nAgain splite the reshape array:\n",splitArr1)

#split vertically
vSplitarr=np.vsplit(reshapeArr,3)
print("\nVertically splite the reshape array:\n",vSplitarr)

#split horizontally

hSplitArr=np.hsplit(reshapeArr,6)
print(f"\nHorizontally splite the reshape array:\n",hSplitArr)
```

## Trignometry function using Numpy

```python
x=np.arange(0,1,0.1)
print("X is:",x)

y=np.sin(x)
print("Y is:",y)
```

### Graphical Representation

```python
plt.plot(x,y)
plt.title("Trignomety Function sin()")
plt.xlabel("Y-value")
plt.ylabel("X-value")
plt.grid=False
plt.show()
```

```python
x=np.arange(1,10,0.5)
y=np.tan(x)
print(f"X:{x}\nY:{y}\n max value in Y:{np.max(y)}")

plt.plot(x,y)
plt.title("Trignometry Function tan()")
plt.xlabel("X-value")
plt.ylabel("Y-value")
plt.grid=True
plt.show()
```
