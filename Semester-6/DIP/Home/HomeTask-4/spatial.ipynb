# Implementation of Spatial Domain Image Enhancement Techniques in Python

## This Python program demonstrates three fundamental spatial domain image enhancement techniques using OpenCV and NumPy:

## Negative Transformation – Inverts pixel values to highlight details in brighter regions.

## Log Transformation – Expands darker pixel values while compressing higher intensities, useful for images with large dynamic ranges.

## Gamma Correction – Adjusts overall image brightness using a power-law function, enhancing either dark or bright regions depending on the gamma value.

## The results are visualized using Matplotlib for direct comparison with the original image.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load image in grayscale
img = cv2.imread("input.jpg", 0)

# -------- 1. Negative Transformation --------
negative = 255 - img

# -------- 2. Log Transformation --------
# Convert image to float for precision
c = 255 / np.log(1 + np.max(img))
log_transformed = c * (np.log(img + 1))
log_transformed = np.array(log_transformed, dtype=np.uint8)

# -------- 3. Gamma Correction --------
gamma = 0.5   # < 1 brightens, > 1 darkens
gamma_corrected = np.array(255 * ((img / 255) ** gamma), dtype=np.uint8)

# -------- Display Results --------
titles = ['Original Image', 'Negative', 'Log Transformation', 'Gamma Correction']
images = [img, negative, log_transformed, gamma_corrected]

for i in range(4):
    plt.subplot(2, 2, i + 1)
    plt.imshow(images[i], cmap='gray')
    plt.title(titles[i])
    plt.axis('off')

plt.show()
```
