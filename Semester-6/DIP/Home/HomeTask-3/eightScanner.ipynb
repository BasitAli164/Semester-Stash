## Stitching Eight Scanned Image Parts into a Single Image using Python PIL

### This Python script uses the Pillow (PIL) library to merge eight equal-sized scanned image parts into a single stitched image. The script assumes the images are arranged in a 2-row × 4-column grid (four images in the top row and four in the bottom row). It automatically creates a new canvas large enough to hold all pieces, then pastes each scanned part into its correct position. Finally, it saves the stitched result as a single high-resolution image file.

```python
from PIL import Image 
```

```python


# Load the 8 scanned images
top_left1 = Image.open("top_left.jpg")
top_left2 = Image.open("top_left.jpg")
top_right1 = Image.open("top_right.jpg")
top_right2 = Image.open("top_right.jpg")

bottom_left1 = Image.open("bottom_left.jpg")
bottom_left2 = Image.open("bottom_left.jpg")
bottom_right1 = Image.open("bottom_right.jpg")
bottom_right2 = Image.open("bottom_right.jpg")

# Get width and height of one scanned piece
width, height = top_left1.size

# Create new image for final stitched result (4 columns × 2 rows)
final_image = Image.new('RGB', (width * 4, height * 2))

# Paste images row by row
# Top row
final_image.paste(top_left1, (0, 0))
final_image.paste(top_left2, (width, 0))
final_image.paste(top_right1, (width * 2, 0))
final_image.paste(top_right2, (width * 3, 0))

# Bottom row
final_image.paste(bottom_left1, (0, height))
final_image.paste(bottom_left2, (width, height))
final_image.paste(bottom_right1, (width * 2, height))
final_image.paste(bottom_right2, (width * 3, height))

# Save final stitched image
final_image.save("stitched_image_8parts.jpg")

print("8-part image stitched successfully!")
```
