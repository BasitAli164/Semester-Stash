```python
import cv2

# Step 1: Read the image from local storage
image = cv2.imread("img.jpeg")  # Replace with your file name/path

if image is None:
    print("Error: Image not found!")
    exit()

# Step 2: Convert to different color formats
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)       # Grayscale
# RGB Color Space

# Step 3: Display all images
cv2.imshow("Original Image", image)
cv2.imshow("Grayscale Image", gray_image)


# Step 4: Wait for a key press and close windows
cv2.waitKey(0)
cv2.destroyAllWindows()
```
