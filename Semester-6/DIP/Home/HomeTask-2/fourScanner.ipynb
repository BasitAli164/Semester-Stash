# Stitching Four Scanned Image Quadrants into a Single Image using Pillow

### This Python script uses the Pillow library to combine four separate scanned image quadrants (top_left, top_right, bottom_left, bottom_right) into one complete stitched image. It creates a new blank canvas with double the width and height of a single quadrant, pastes each scanned image into its correct position, and saves the final merged output as stitched_image.jpg.

```python
from PIL import Image
```

```python


# Load the 4 scanned images
top_left = Image.open("top_left.jpg")
top_right = Image.open("top_right.jpg")
bottom_left = Image.open("bottom_left.jpg")
bottom_right = Image.open("bottom_right.jpg")

# Get the width and height of each quadrant
width, height = top_left.size

# Create a new image to hold the final stitched image
final_image = Image.new('RGB', (width * 2, height * 2))

# Paste each scanned image into the correct position
final_image.paste(top_left, (0, 0))
final_image.paste(top_right, (width, 0))
final_image.paste(bottom_left, (0, height))
final_image.paste(bottom_right, (width, height))

# Save the final stitched image
final_image.save("stitched_image.jpg")

print("Image stitched successfully!")
```
