```python
# ----------------------------------------------
# Import necessary libraries
# ----------------------------------------------

import pandas as pd  
# Pandas is a data analysis library.
# It allows us to read and work with structured data (CSV, Excel, etc.)
# in a DataFrame format (like a table with rows & columns).
# Here, it is used to read the email dataset and manipulate data easily.

from sklearn.feature_extraction.text import CountVectorizer  
# Converts text into numerical form by tokenizing words and counting how
# many times each word appears. This creates a "bag-of-words" representation
# that can be used by machine learning algorithms.

from sklearn.model_selection import train_test_split  
# Splits the dataset into training and testing parts.
# Training data → used to train the model.
# Testing data → used to evaluate how well the model generalizes to new data.

from sklearn.naive_bayes import MultinomialNB  
# Multinomial Naive Bayes classifier.
# Well-suited for text classification tasks where features represent word counts or frequencies.
# It works on probability principles and assumes independence between words.

from sklearn.pipeline import make_pipeline  
# Combines multiple processing steps into a single object.
# For example, it chains CountVectorizer (text to numbers) and MultinomialNB (classification).
# With a pipeline, we can directly call .fit() and .predict() without doing each step manually.


# ----------------------------------------------
# Import dataset
# ----------------------------------------------

path = '/content/drive/MyDrive/email_classification_dataset.csv'  
# Path to the dataset file stored in Google Drive (or local path if running elsewhere).

df = pd.read_csv(path)  
# Reads the CSV file into a Pandas DataFrame.
# The dataset is expected to have two main columns: "email" (text) and "label" (spam/ham/etc.).

print(df.head())  
# Displays the first 5 rows of the dataset for a quick preview.

print(f"Shape of Dataset is: {df.shape}")  
# Prints the shape of the dataset → (rows, columns).
# Rows = number of emails, Columns = features/labels in the dataset.


# ----------------------------------------------
# Extract features and labels
# ----------------------------------------------

X = df['email']  
# Input features (independent variable) → the actual email text.

y = df['label']  
# Output labels (dependent variable) → the category each email belongs to (e.g., spam/ham).


# ----------------------------------------------
# Split dataset into training & testing
# ----------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=40
)  
# Splits the data into training and testing sets:
# - 70% training, 30% testing (because test_size=0.3).
# - random_state=40 ensures the split is reproducible.
# This prevents overfitting and allows evaluation on unseen data.


# ----------------------------------------------
# Create pipeline (Vectorizer + Classifier)
# ----------------------------------------------

model = make_pipeline(CountVectorizer(), MultinomialNB())  
# Step 1: CountVectorizer → convert raw emails into word count vectors.
# Step 2: MultinomialNB → train Naive Bayes classifier on these vectors.
# Pipeline ensures both steps are applied together automatically.


# ----------------------------------------------
# Train the model
# ----------------------------------------------

model.fit(X_train, y_train)  
# Trains the pipeline:
# - CountVectorizer learns vocabulary from training emails.
# - MultinomialNB learns probability distribution of words for each label (spam/ham/etc.).


# ----------------------------------------------
# Test on new unseen emails
# ----------------------------------------------

test_emails = [
    # Example emails we want to classify
    "Congratulations! You won a free lottery ticket.",  # Looks like spam
    "Limited time offer, claim your prize now!",         # Looks like spam
    "Meeting scheduled for tomorrow at 10 AM.",         # Likely ham (work email)
    "Your invoice for last month is attached.",         # Likely ham (business email)
]

results = model.predict(test_emails)  
# Predicts the label for each email.
# - CountVectorizer transforms test emails into word count vectors.
# - MultinomialNB uses the learned model to classify them.


# ----------------------------------------------
# Display results
# ----------------------------------------------

print("\n--- Predictions on New Emails ---")  
# Header to separate output.

for email, label in zip(test_emails, results):
    print("=====================================")
    print(f"Email Text   : {email}")   # The email text
    print(f"Prediction   : {label}")   # The predicted category (spam/ham/etc.)
```

Import libraries → For data handling, text processing, splitting, classification, and chaining steps.

Load dataset → Reads CSV file of emails and their labels.

Extract features and labels → X = email text, y = labels.

Split dataset → 70% for training, 30% for testing.

Pipeline → Automatically applies CountVectorizer + Naive Bayes.

Train model → Learns how to distinguish spam vs ham from training data.

Predict on new emails → Classifies unseen examples.

Show results → Prints predictions clearly.
