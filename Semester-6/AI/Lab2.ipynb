```python
# ----------------------------------------------
# Import necessary libraries
# ----------------------------------------------

from sklearn.feature_extraction.text import CountVectorizer  
# CountVectorizer converts text into numerical form by tokenizing (splitting) words 
# and then counting how many times each word appears in the text.  
# Example: ["win a car", "win cash"] → {"win":2, "a":1, "car":1, "cash":1}.  
# This is needed because ML models cannot directly work with raw text.

from sklearn.naive_bayes import MultinomialNB  
# Multinomial Naive Bayes is a machine learning algorithm that is 
# especially effective for text classification tasks where data is based on 
# word counts or word frequencies.  
# It assumes words appear independently and calculates the probability 
# of a message belonging to a category (spam/ham/important/etc.).

from sklearn.pipeline import make_pipeline  
# A pipeline chains multiple processing steps together into one object.  
# For example, here it will first convert text into word counts (CountVectorizer), 
# and then apply the Naive Bayes classifier (MultinomialNB).  
# This simplifies training and prediction (fit and predict happen in sequence).


# ----------------------------------------------
# Dummy training dataset
# ----------------------------------------------

data = [
    ("win a free iphone ", "spam"),                   # Example of spam message
    ("Your Invoice for last month", "ham"),           # Ham = normal/legit email
    ("meeting scheduled at 10 Am", "ham"),            # Legit meeting reminder
    ("congratulation , you won a lottery", "spam"),   # Spam example
    ("Win a brand new car", "spam"),                  # Spam again
    ('Le projet est deead line  in Tucky!', 'important'), # Important project reminder (different language mix)
    ("This AI Course is just for your!", 'Public'),   # Public/advertisement example
]
# Each entry is a tuple: (email text, label).  
# Labels: "spam", "ham", "important", "Public".  
# This dataset is very small and only for demonstration.


# ----------------------------------------------
# Separate features (X) and labels (y)
# ----------------------------------------------

Texts, label = zip(*data)  
# The zip(*) trick "unzips" the list of tuples into two tuples:  
# - Texts = ("win a free iphone", "Your Invoice...", ...)  
# - label = ("spam", "ham", "ham", "spam", ...)  
# This way, we separate input features (email text) and output targets (labels).


# ----------------------------------------------
# Create a pipeline (CountVectorizer + Naive Bayes)
# ----------------------------------------------

model = make_pipeline(CountVectorizer(), MultinomialNB())  
# Pipeline steps:  
# 1. CountVectorizer() → converts email text into a bag-of-words numeric form.  
# 2. MultinomialNB() → trains a classifier based on those word counts.  
# The pipeline ensures both steps run automatically in sequence.


# ----------------------------------------------
# Train the model
# ----------------------------------------------

model.fit(Texts, label)  
# Fits (trains) the pipeline using training data.  
# - CountVectorizer learns the vocabulary from training texts.  
# - MultinomialNB learns probabilities of words belonging to each label.  
# Example: "win" appears often in spam → higher probability of spam if "win" is present.


# ----------------------------------------------
# Test dataset (new unseen emails)
# ----------------------------------------------

test_mail = [
    "Win a Brand New Car",         # Likely spam (word "Win" + "Car")
    "Project deadline is Today!",  # Likely important (project/deadline context)
    "I will kill you any time!....", # Dangerous/unknown (not in training data, model may misclassify)
    "meeting scheduled at 10 Am",   # Should be 'ham' (it appeared in training data)
    "This AI Course is just for your!", # Likely 'Public'
    "Le projet est deead line  in Tucky!", # Likely 'important'
    "Limited time offer for you"     # Likely 'spam' (marketing phrase)
]


# ----------------------------------------------
# Make predictions
# ----------------------------------------------

prediction = model.predict(test_mail)  
# Uses the trained model to predict labels for each test email.  
# Steps internally:  
# - CountVectorizer converts test text to word counts (based on training vocabulary).  
# - MultinomialNB applies learned probabilities to classify each message.


# ----------------------------------------------
# Display results
# ----------------------------------------------

for email, label in zip(test_mail, prediction):
    print("=====================================")
    print(f"Email Text   : {email}")    # Print the test email
    print(f"Prediction   : {label}")    # Print the predicted category
```

Import Tools → For text vectorization, classification, and chaining.

Training Data → Example emails with labels.

Separation → Texts = input, Labels = output.

Pipeline → Converts text → word counts → classifier.

Training → Model learns spam vs ham vs important vs public.

Prediction → New emails classified.

Result → Prints predictions.
