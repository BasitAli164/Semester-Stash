```python
# Student Marks Prediction using AI (Simple Example)
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Step 1: Data (Hours studied vs Marks obtained)
hours = [[1], [2], [3], [4], [5], [6], [7], [8]]
marks = [35, 40, 50, 55, 65, 70, 80, 85]

# Step 2: Model banana
model = LinearRegression()

# Step 3: Model train karna
model.fit(hours, marks)

# Step 4: Prediction (AI ko test karna)
predicted_marks = model.predict([[6]])

# Step 5: Output
print("Agar student 6 ghantay padhe to estimated marks =", round(predicted_marks[0], 2))

# Step 6: Visualization (Graph)
plt.scatter(hours, marks, color='blue', label='Actual Data (Students)')
plt.plot(hours, model.predict(hours), color='red', label='AI Learned Line')
plt.scatter(9, predicted_marks, color='green', label='Prediction (9 hours)')
plt.title("Student Marks Prediction using AI")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.legend()
plt.show()
```
