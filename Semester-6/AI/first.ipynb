```python
import numpy as np
from sklearn.linear_model import LinearRegression

# Input data (2D array with 4 values)
X = np.array([[1], [2], [3], [4]])   # shape (4,1)

# Output is just 2 * X
y = np.array([2, 4, 6, 8])

# Create and train the model
model = LinearRegression()
model.fit(X, y)

# Test prediction
test_val = np.array([[80]])
prediction = model.predict(test_val)

print("Coefficient (slope):", model.coef_)
print("Intercept:", model.intercept_)
print(f"Prediction for {test_val}:", prediction)
```
