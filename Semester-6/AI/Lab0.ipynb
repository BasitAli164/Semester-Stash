# Work With LinearRegression Model to predict unseen numeric values

```python
# ----------------------------------------------
# Import necessary libraries
# ----------------------------------------------

import numpy as np  
# NumPy is a library for numerical computing in Python.
# It provides fast array and matrix operations. Here, it's used
# to create and manipulate numerical data (input X and output y).

from sklearn.linear_model import LinearRegression  
# 'sklearn' (scikit-learn) is a machine learning library in Python.
# 'linear_model.LinearRegression' is a class that implements linear regression,
# a supervised learning algorithm used for predicting continuous values.
# We import this model because we want to fit a line (y = mx + c)
# to our training data and then make predictions.


# ----------------------------------------------
# Input data preparation
# ----------------------------------------------

X = np.array([[1], [2], [3], [4]])  
# Input (independent variable) dataset.
# 'X' is a 2D array (shape: 4 rows × 1 column), where each row
# is a data point (feature value). Example: X = [1, 2, 3, 4].
# In sklearn, input features must be in 2D form, hence the double brackets.

y = np.array([2, 4, 6, 8])  
# Output (dependent variable) dataset.
# 'y' is a 1D array, containing the target values corresponding to X.
# The relationship is simple: y = 2 * X (linear relationship).


# ----------------------------------------------
# Create and train the model
# ----------------------------------------------

model = LinearRegression()  
# Create an instance of the LinearRegression model.
# At this point, the model is untrained (no slope/intercept known yet).

model.fit(X, y)  
# Train (fit) the model on the data (X → y).
# The algorithm finds the best-fitting line (slope and intercept)
# that minimizes the error between predicted and actual y values.
# In this case, it will learn slope ≈ 2 and intercept ≈ 0.


# ----------------------------------------------
# Make a prediction
# ----------------------------------------------

test_val = np.array([[80]])  
# A new test input value for prediction.
# Again, it must be 2D (array of arrays) because sklearn expects this format.
# Here, X = 80 (input feature value).

prediction = model.predict(test_val)  
# The model predicts the output (y value) corresponding to test_val.
# Using the learned linear relationship: y ≈ slope * X + intercept.
# For input 80, prediction ≈ 2 * 80 = 160.


# ----------------------------------------------
# Extract model parameters and convert arrays to scalars
# ----------------------------------------------

slope = model.coef_[0]  
# The learned slope (coefficient) of the line.
# 'model.coef_' returns an array, so we take the first element.
# In this case, slope = 2.

intercept = model.intercept_  
# The learned intercept (bias term).
# It’s the point where the line crosses the Y-axis (when X = 0).
# In this case, intercept = 0.

predicted_value = prediction[0]  
# The prediction result is also in array form.
# Extract the first (and only) element to convert it into a scalar.


# ----------------------------------------------
# Print the results
# ----------------------------------------------

print("Coefficient (slope):", slope)  
# Displays the slope of the fitted line.

print("Intercept:", intercept)  
# Displays the intercept of the fitted line.

print(f"Prediction for {test_val[0][0]}:", predicted_value)  
# Displays the predicted y-value for the test input X = 80.
# Expected output: Prediction for 80: 160.
```
