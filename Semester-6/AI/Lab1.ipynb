```python
from sklearn.feature_extraction.text import CountVectorizer # tokonize the text and count the  word appear
from sklearn.naive_bayes import MultinomialNB # it is the ml alogorithm to classified well suited for word count 
from sklearn.pipeline import make_pipeline #  It prepare a chain to fit and  predict
```

```python
data=[
    ("win a free iphone ","spam"),
    ("Your Invoice for last month","ham"),
    ("meeting scheduled at 10 Am","ham"),
    ("congratulation , you won a lottery","spam"),
    ("Win a brand new car", "spam"),
    ('Le projet est deead line  in Tucky!','important'),
    ("This AI Course is just for your!", 'Public'),
    
    ]
```

```python
# Separte data
Texts,label=zip(*data) # * unpack data and whole groups label and text accros all tuples

#create a piplein
model=make_pipeline(CountVectorizer(),MultinomialNB())

#Train the model 
model.fit(Texts,label)


# Test emial

test_mail=["Win a Brand New Car",
           "Project deadline is Today!",
           "I will kill you any time!....",
           "meeting scheduled at 10 Am",
           "This AI Course is just for your!",
           "Le projet est deead line  in Tucky!",
           "Limited time offer for you"]


# Predict

prediction=model.predict(test_mail)


#result

for email, label in zip(test_mail,prediction):
    print(f"Email.......Prediction:{label}")
```
