```python
import gradio as gr

# Define the function for the Love Compatibility Calculator
def calculate_love_score(name1, name2):
    # Merge the inputs to one string
    name = name1 + name2
    # Convert the inputs to lower case
    name = name.lower()
    
    # Initialize the keywords to check
    true_keyword = "true"
    love_keyword = "love"

    # Initialize the counters to 0 
    true_count = 0
    love_count = 0

    # Count occurrences of characters from keywords
    for input_char in name:
        if input_char in true_keyword:
            true_count += 1
        if input_char in love_keyword:
            love_count += 1  

    # Concatenate the true score and love score as a string
    final_score = int(str(true_count) + str(love_count))
    
    # Determine compatibility level based on the score
    if final_score < 20:
        compatibility = "Very Poor"
    elif final_score < 30:
        compatibility = "Very Low"
    elif final_score < 40:
        compatibility = "Good"
    elif final_score < 50:
        compatibility = "Very Good"
    else:
        compatibility = "Excellent"

    # Return the results
    return f"Compatibility Percentage: {final_score}%\nCompatibility Level: {compatibility}"

# Create a Gradio interface
interface = gr.Interface(
    fn=calculate_love_score,  # Function to use
    inputs=[
        gr.Textbox(label="Partner 1 Name", placeholder="Enter the first partner's name"),
        gr.Textbox(label="Partner 2 Name", placeholder="Enter the second partner's name")
    ],
    outputs="text",  # The output will be text
    title="Love Compatibility Calculator",
    description="Enter the names of two people to calculate their love compatibility based on the keywords 'TRUE' and 'LOVE'."
)

# Launch the Gradio app with a public link
interface.launch(share=True)
```
